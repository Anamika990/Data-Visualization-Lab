````python
# Install required packages
!pip install -U google-generativeai clint beautifulsoup4 requests

# Imports
import os
import re
import requests
from bs4 import BeautifulSoup
import google.generativeai as genai
from clint.textui import colored  # For colored terminal output
import time  # To measure execution time

# Step 1: Clean text function
def clean_text(text):
    """
    Cleans the input text using regular expressions:
    - Removes HTML tags
    - Removes URLs
    - Removes special characters
    - Normalizes whitespace
    """
    text = re.sub(r'<[^>]+>', '', text)  # HTML tags
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)  # URLs
    text = re.sub(r"[^a-zA-Z0-9.,;!?'\s]", '', text)  # Special characters
    text = re.sub(r'\s+', ' ', text).strip()  # Normalize whitespace
    return text

# Step 2: Extract & clean data from URLs using BeautifulSoup
def extract_and_clean_data_from_urls(url_list):
    documents = []
    for url in url_list:
        try:
            response = requests.get(url)
            soup = BeautifulSoup(response.content, 'html.parser')
            raw_text = soup.get_text()
            cleaned_text = clean_text(raw_text)
            documents.append(cleaned_text)
        except Exception as e:
            print(colored.red(f"❌ Error loading {url}: {e}"))
    return documents

# Step 3: Setup Gemini API
def setup_gemini(api_key=None):
    if api_key is None:
        api_key = os.getenv("GEMINI_API_KEY")
    if api_key:
        genai.configure(api_key=api_key)
        print(colored.green("Gemini API is configured successfully."))
    else:
        print(colored.red("API Key is not provided."))

# Step 4: Generate content using Gemini
def generate_content_with_gemini(prompt_text, model_name="models/gemini-1.5-flash"):
    try:
        model = genai.GenerativeModel(model_name)
        response = model.generate_content(prompt_text)
        return response.text
    except Exception as e:
        return f"Error during generation: {e}"


# Step 5: Initialize Gemini API

api_key = "AIzaSyAAFVgwskVBVTgPLHAxp6feBpNfjulHXGw"  # my api key
setup_gemini(api_key)


# Step 6: creating system prompt
system = """ Do not use code block formatting like ```json.
You are a network graph maker tasked with analyzing the relationships involving top leaders in the world. Your job is to process the provided context chunk 
and extract an ontology of terms that represent key entrepreneurs, their associated entities, and all kinds of relationships present in the context.

**Guidelines for Extraction:**

1. **Identify Key Entrepreneurs and Related Terms**:
   - Extract key entrepreneurs and related concepts such as:
     - Companies, organizations, or industries they are associated with.
     - Collaborators, partners, rivals, or competitors.
     - Key innovations, achievements, or milestones.
     - Locations, events, or time periods relevant to their actions.

2. **Identify Relationships**:
   - Extract all types of relationships between entrepreneurs and other entities (or between entities themselves).
   - Relationships can include:
     - Professional roles or associations.
     - Business partnerships, collaborations, or rivalries.
     - Innovations or contributions to industries.
     - Personal connections or influences.
     - Historical events or shared milestones.

3. **Define Relationships**:
   - Clearly specify the nature of each relationship in simple and concise terms.
   - Relationships should convey meaningful connections relevant to the context.

**Response Format**:
- Provide your output **strictly as a list of JSON objects**. No additional text, descriptions, tags or comments are allowed.
- Each object should include the following fields:
  - `"node_1"`: The first entity in the relationship (can be a person, organization, or concept).
  - `"node_2"`: The second entity in the relationship.
  - `"edge"`: A concise sentence describing the relationship between `node_1` and `node_2`.

**Example Output**:
[
   {
       "node_1": "Elon Musk",
       "node_2": "SpaceX",
       "edge": "Elon Musk founded SpaceX to revolutionize space exploration."
   },
   {
       "node_1": "Steve Jobs",
       "node_2": "Apple Inc.",
       "edge": "Steve Jobs co-founded Apple Inc., a leading tech company."
   },
   {
       "node_1": "Mark Zuckerberg",
       "node_2": "Sheryl Sandberg",
       "edge": "Sheryl Sandberg worked closely with Mark Zuckerberg as COO of Facebook."
   },
   {
       "node_1": "Jeff Bezos",
       "node_2": "Blue Origin",
       "edge": "Jeff Bezos founded Blue Origin to focus on space exploration."
   }
]
**Important Note**:
- Do not use code block formatting like ```json.
- Always respond exclusively in JSON format. Any deviation from the JSON structure or inclusion of additional text will not be accepted.
- Output must be a valid JSON array of objects without any surrounding text.
Please provide the context containing information about entrepreneurs and their relationships for analysis.
"""

# Track time is taken for generation
start_time = time.time()


# Generate and display the content
generated_content = generate_content_with_gemini(system)


# Track time taken and calculate duration
end_time = time.time()
elapsed_time = end_time - start_time


# Display generated content and the time taken
print(colored.cyan("\n Generated Content:\n"), colored.white(generated_content))
print(colored.green(f"\n Time taken for data extraction: {elapsed_time:.2f} seconds"))


# Save the generated content as JSON
import json

# Function to clean the response before attempting to load it as JSON
def clean_json_response(response):
    # Strip unwanted markdown formatting (e.g., ```json or similar)
    cleaned_response = response.strip('```json').strip('```').strip()
    # Remove any additional markdown-like formatting or characters that might remain
    cleaned_response = re.sub(r'```.*```', '', cleaned_response).strip()  # Regex for any markdown
    return cleaned_response


# Check if the response seems to be in JSON format
def is_valid_json(response):
    try:
        # Clean the response first
        cleaned_response = clean_json_response(response)
        
        # Attempt to load the cleaned response as JSON
        json.loads(cleaned_response)
        return True
    except ValueError:
        return False


# Only attempt to parse if the content is valid JSON
if is_valid_json(generated_content):
    # Clean the response before parsing
    cleaned_content = clean_json_response(generated_content)
    result_json = json.loads(cleaned_content)  # Convert the string response to JSON

    with open('Nodes_and_edges.json', 'w') as json_file:
        json.dump(result_json, json_file, indent=4)

    print("Saved the extracted data to 'Nodes_and_edges.json'", "green")
else:
    print("Error: The generated content is not a valid JSON response.")
    # Print the raw response for debugging purposes
    print("Raw response received from the API:", colored.white(generated_content))
````

````python
import json
from gephistreamer import graph
from gephistreamer import streamer


# Connect to Gephi Streaming API
stream = streamer.Streamer(streamer.GephiWS(hostname="localhost", port=8080, workspace="workspace1"))


# Load JSON data
with open('Nodes_and_edges.json', 'r') as file:
    results = json.load(file)


# Process each entry in the JSON
for res in results:
    try:
        # Extract values
        node_1 = res['node_1']
        node_2 = res['node_2']
        edge_label = res['edge']

        # Create nodes
        node_a = graph.Node(node_1, custom_property=1)
        node_b = graph.Node(node_2, custom_property=2)

        # Add nodes to stream
        stream.add_node(node_a)
        stream.add_node(node_b)

        # Create and add edge
        edge_ab = graph.Edge(node_a, node_b, directed=True, label=edge_label)
        stream.add_edge(edge_ab)

    except Exception as e:
        print('Buggy JSON object:', e, res)
````
